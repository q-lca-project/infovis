```python
"""
-------------------------------------------------------------------------------
Initial visual exploration of selected neighbourhood scenarios to identify
relevant differences, patterns and potential insights for further analysis.

The scenarios are defined by six factors:
- Building type
- Density
- Road layout
- Construction
- Heat supply
- Energy efficiency level

Scenario comparisons:

1. Block vs. Single-family house
   - Block: Block / High density / Curve / Wood / Local energy /  30 kWh
   - Single-family: Single / Low density / Wood / Curve / Local energy / 30 kWh

2. Row house vs. Single-family house
   - Row house: Row / Low density / Grid / Mineral / Local energy / 15 kWh
   - Single-family: Single / Low density / Grid / Mineral / Local energy / 15 kWh 
   - Optimised variant: High density / Wood

3. Dynamic vs. Static QLCA
   - Line house: Line / Medium density / Curve / Mineral / Local energy / 30 kWh 

------------------------------------------------------------------------------
"""
```

```python
"""
------------------------------------------------------------------------------
Libraries
------------------------------------------------------------------------------
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

import plotly.graph_objects as go
```

```python
# show all columns
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

```python
"""
------------------------------------------------------------------------------
Import data
------------------------------------------------------------------------------
"""


df = pd.read_csv("../data/csv/QLCA_data.csv")

df.head()
```

```python
"""
------------------------------------------------------------------------------
Function: Filter dataset based on the 6 factors
------------------------------------------------------------------------------
"""

def get_scenario(df, scenario):
    mask = pd.Series(True, index=df.index)

    for col, value in scenario.items():
        mask &= df[col] == value

    return df.loc[mask]
```

```python
"""
------------------------------------------------------------------------------
Scenario1: Block
------------------------------------------------------------------------------
"""

S1_BW = {
    "Typ": "BW",
    "Bauart": "Holz",
    "Energiebedarf": "30 kWh",
    "Versorgungsart": "Nahwärme kalt",
    "Dichte": "hoch",
    "Erschließung (Sied.)": "Schlaufe"
}
```

```python
df_S1_BW = get_scenario(df, S1_BW)

df_S1_BW.head()
```

```python
row = df_S1_BW.iloc[0]

# Percent
total = row["GWP"]

GWP_S1_BW_pct = {
    "KG300": row["GWP für Bauwerk (KG 300)"] / total,
    "KG400": row["GWP für Haustechnik (KG 400)"] / total,
    "KG500": row["GWP für gesamte Geb.- Erschliessung (KG 500)"] / total,
    "B6": row["GWP für Wärmeversorgung (B6)"] / total
}
```

```python
kg_colors = {
    "KG300": "#3A86FF", 
    "KG400": "#FD2B3B",
    "KG500": "#10B61D",
    "B6":    "#FFB20B"
}
```

```python
# Percent

fig = go.Figure()

# KG300
fig.add_trace(go.Bar(
    x=["Scenario"],
    y=[GWP_S1_BW_pct["KG300"]],
    name="KG300",
    marker_color=kg_colors["KG300"],
    width=0.3,
    marker_line_color="white",
    marker_line_width=1
))

# KG400
fig.add_trace(go.Bar(
    x=["Scenario"],
    y=[GWP_S1_BW_pct["KG400"]],
    name="KG400",
    marker_color=kg_colors["KG400"],
    width=0.3,
    marker_line_color="white",
    marker_line_width=1
))


# KG500 
fig.add_trace(go.Bar(
    x=["Scenario"],
    y=[GWP_S1_BW_pct["KG500"]],
    name="KG500",
    marker_color=kg_colors["KG500"],
    width=0.3,
    marker_line_color="white",
    marker_line_width=1
))

# B6
fig.add_trace(go.Bar(
    x=["Scenario"],
    y=[GWP_S1_BW_pct["B6"]],
    name="B6",
    marker_color=kg_colors["B6"],
    width=0.3,
    marker_line_color="white",
    marker_line_width=1
))

fig.update_layout(
    barmode="stack",
    title="GWP Breakdown in percent – Scenario 1 Block",
    yaxis_title="Share of total GWP",
    legend_title="Components",
    height=1000,
    width=1100,
    legend=dict(orientation="h", y=-0.2),
    legend_traceorder="normal",
    yaxis=dict(
        tickformat=".0%",
        range=[0, 1]
    )
)

fig.show()
```
