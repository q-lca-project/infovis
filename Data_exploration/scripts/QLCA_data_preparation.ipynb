```python
"""
------------------------------------------------------------------------------
QLCA Data Preparation

Preparation of the QLCA dataset used for further analysis.

The data is extracted from the Excel file
`Q-LCA_FD3_Bilanzierungstool.xlsx`, using Tables 4, 8 and 9.

The workflow includes:
- identification of relevant tables and variables
- extraction and combination of the required data
- selection of relevant columns
- data cleaning and preparation for further analysis
------------------------------------------------------------------------------
"""
```

```python
"""
------------------------------------------------------------------------------
Libraries
------------------------------------------------------------------------------
"""

import pandas as pd
import numpy as np
```

```python
"""
------------------------------------------------------------------------------
Table 8: Abs_Kod_Matrix_Siedlung / 972 Neighbourhood scenarios
------------------------------------------------------------------------------
"""

# Show all columns
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

# 6ha-Table
df_8_Matrix = pd.read_excel("../data/raw/Q-LCA_FD3_Bilanzierungstool.xlsx", sheet_name="8 - Abs_Kod_Matrix_Siedlung", skiprows =1)
df_8_Matrix.columns = df_8_Matrix.columns.str.replace("\n", " ").str.strip()

df_8_Matrix.head()
```

```python
# Delete all Nan row
df_8_Matrix = df_8_Matrix.dropna(how="all")

# Delete all Nan columns
df_8_Matrix = df_8_Matrix.dropna(axis=1, how="all")

df_8_Matrix.head()
```

```python
# Clean tabs and spaces from column names
df_8_Matrix.columns = (
    df_8_Matrix.columns
    .str.replace("\n", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

df_8_Matrix.head()
```

```python
# Rename unnamed columns
df_8_Matrix = df_8_Matrix.rename(columns={
    "Geschoss-Code": "Geschoss-Code_1",
    "Unnamed: 5": "Geschoss-Code_2",
    "Unnamed: 6": "Geschoss-Code_3"
})

# Combine
df_8_Matrix["Geschoss-Code"] = (
    df_8_Matrix[["Geschoss-Code_1", "Geschoss-Code_2", "Geschoss-Code_3"]]
    .fillna("")
    .astype(str)
    .agg("-".join, axis=1)
)

# Drop columns
df_8_Matrix = df_8_Matrix.drop(columns=[
    "Geschoss-Code_1",
    "Geschoss-Code_2",
    "Geschoss-Code_3"
])


df_8_Matrix.head()
```

```python
print(df_8_Matrix.columns.tolist())
```

```python
relevant_cols = [
    "Siedlungs-Code",
    "Flächen-Code",
    "Typ",
    "Bauart",
    "Energiebedarf",
    "Versorgungsart",
    "Dichte",
    "Erschließung (Segm.)",
    "Erschließung (Sied.)",
    "GWP",
    "GWP ohne Strom",
    "GWP für Bauwerk (KG 300)",
    "310 Erdaushub Keller",
    "322 Bodenplatte",
    "325 Bodenbelag",
    "331 Außenwand",
    "334 Außentüren",
    "339 Kellerwände",
    "334 Fenster",
    "341 Innenwände tragend (inkl. Aufzugschacht)",
    "342 Innen wände",
    "344 Innen türen",
    "351 Decken",
    "351 Treppe",
    "361 Dach",
    "GWP für Haus- technik (KG 400)",
    "410 Abwasser, Trinkwasser",
    "420 Wärme- versorgung",
    "430 Lüftung",
    "440 Strom",
    "460 Aufzug",
    "GWP für gesamte Geb.- Erschliessung (KG 500)",
    "500 Quartiers strom",
    "500 Quartierswasserleitung",
    "510 Erdaushub Leitungen",
    "510 Erdaushub Garagen",
    "521 Gehweg Radweg",
    "522 Straße",
    "524 Stell -plätze",
    "524 Hoch garagen",
    "524 Tief garagen",
    "Summe Betrieb & Nutzstrom",
    "GWP für Wärmeversorgung (B6)",
    "GWP für Nutzstrom"
]

df_8 = df_8_Matrix[relevant_cols].copy()

df_8.head()
```

```python
# Rename columns
df_8 = df_8.rename(columns={

    # KG 400
    "GWP für Haus- technik (KG 400)": "GWP für Haustechnik (KG 400)",

    # KG500
    "500 Quartiers strom": "500 Quartiersstrom"
}
)
```

```python
print(df_8["GWP"].dtype)
print(df_8["GWP für Bauwerk (KG 300)"].dtype)
print(df_8["GWP für Haustechnik (KG 400)"].dtype)
print(df_8["GWP für gesamte Geb.- Erschliessung (KG 500)"].dtype)
print(df_8["Summe Betrieb & Nutzstrom"].dtype)
```

```python
df_8["GWP"] = df_8["GWP"].round(0).astype("Int64")

df_8["GWP für Bauwerk (KG 300)"] = df_8["GWP für Bauwerk (KG 300)"].astype("float")

df_8["GWP für Haustechnik (KG 400)"] = df_8["GWP für Haustechnik (KG 400)"].round(1).astype("Float64")

df_8["GWP für gesamte Geb.- Erschliessung (KG 500)"] = df_8["GWP für gesamte Geb.- Erschliessung (KG 500)"].round(0).astype("Int64")

df_8["Summe Betrieb & Nutzstrom"]= df_8["Summe Betrieb & Nutzstrom"].round(0).astype("Int64")
```

```python
"""
------------------------------------------------------------------------------
Table 9: ALL_NGF_Kod_Matrix_Siedlung / GWP per qm
------------------------------------------------------------------------------
"""

# 6ha-Table
df_9_Matrix = pd.read_excel("../data/raw/Q-LCA_FD3_Bilanzierungstool.xlsx", sheet_name="9 - ALL_NGF_Kod_Matrix_Siedlung", skiprows =1)
df_9_Matrix.columns = df_9_Matrix.columns.str.replace("\n", " ")

df_9_Matrix.head()
```

```python
# Clean tabs and spaces from column names
df_9_Matrix.columns = (
    df_9_Matrix.columns
    .str.replace("\n", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
df_9_Matrix.head()
```

```python
relevant_cols = [
    "Siedlungs-Code",
    "Flächen-Code",
    "GWP/ qm Basic + Aushub +Strom"
]

df_9 = df_9_Matrix[relevant_cols].copy()

df_9.head()
```

```python
# Rename column

df_9 = df_9.rename(columns={

    "GWP/ qm Basic + Aushub +Strom": "GWP / m² NRF"

})

df_9.head()
```

```python
# Siedlungs-Code / neighbourhood code
df_4_8["Siedlungs-Code"] = df_4_8["Siedlungs-Code"].str.strip().str.upper()
df_9["Siedlungs-Code"] = df_9["Siedlungs-Code"].str.strip().str.upper()

# Flächen-Code / area code
df_4_8["Flächen-Code"] = df_4_8["Siedlungs-Code"].str.strip().str.upper()
df_9["Flächen-Code"] = df_9["Siedlungs-Code"].str.strip().str.upper()
```

```python
# Merge df_4_8 + df_9

df_QLCA = df_4_8.merge(df_9, on=["Siedlungs-Code", "Flächen-Code"], how= "left")

df_QLCA.head()
```

```python
# Tiefgaragen yes / no?

print(df_QLCA["Flächen Tiefgaragen (brutto) (2)"].dtypes)
print(df_QLCA["Flächen Tiefgaragen (netto) (3)"].dtypes)
print(df_QLCA["Flächen Tiefgaragen (brutto) (2)"].value_counts(dropna=False))
print(df_QLCA["Flächen Tiefgaragen (netto) (3)"].value_counts(dropna=False))
```

```python
# Add column "Tiefgarage"

df_QLCA["Tiefgarage"] = df_QLCA["Flächen Tiefgaragen (brutto) (2)"].notna()

df_QLCA.head()
```

```python
df_QLCA = df_QLCA.rename(columns={
    "524 Tief garagen": "524 Tiefgaragen",
    "342 Innen wände": "342 Innenwände",
    "344 Innen türen": "344 Innentüren",
    "420 Wärme- versorgung": "420 Wärmeversorgung",
    "524 Stell -plätze": "524 Stellplätze",
    "524 Hoch garagen": "524 Hochgaragen",
    "524 Tief garagen": "524 Tiefgaragen",
    "Rad- und Gehweg- flächen (öffentlich)": "Rad- und Gehwegflächen (öffentlich)"
})

df_QLCA.head()
```

```python
"""
------------------------------------------------------------------------------
Verkehrsfläche gesamt
------------------------------------------------------------------------------
"""

df_QLCA["Verkehrsfläche gesamt m²"] = df_QLCA["Straßenflächen (öffentlich)"] + df_QLCA["Rad- und Gehwegflächen (öffentlich)"] + df_QLCA["Oberirdische Stellplatzflächen (öffentlich)"]
```

```python
"""
------------------------------------------------------------------------------
Add columns: GWP per Person
------------------------------------------------------------------------------
"""

denom = df_QLCA["Bewohner Kachel insg. (rechnerisch)"]

# --- GWP ---
df_QLCA["Person_GWP"] = (df_QLCA["GWP"] / denom).round(2)

df_QLCA["Person_GWP (GWP ohne Strom)"] = (
    df_QLCA["GWP ohne Strom"] / denom
).round(2)

# --- KG 300 ---
df_QLCA["Person_GWP (KG 300)"] = (
    df_QLCA["GWP für Bauwerk (KG 300)"] / denom
).round(2)

df_QLCA["Person_GWP (310 Erdaushub Keller)"] = (
    df_QLCA["310 Erdaushub Keller"] / denom
).round(2)

df_QLCA["Person_GWP (322 Bodenplatte)"] = (
    df_QLCA["322 Bodenplatte"] / denom
).round(2)

df_QLCA["Person_GWP (325 Bodenbelag)"] = (
    df_QLCA["325 Bodenbelag"] / denom
).round(2)

df_QLCA["Person_GWP (331 Außenwand)"] = (
    df_QLCA["331 Außenwand"] / denom
).round(2)

df_QLCA["Person_GWP (334 Außentüren)"] = (
    df_QLCA["334 Außentüren"] / denom
).round(2)

df_QLCA["Person_GWP (339 Kellerwände)"] = (
    df_QLCA["339 Kellerwände"] / denom
).round(2)

df_QLCA["Person_GWP (334 Fenster)"] = (
    df_QLCA["334 Fenster"] / denom
).round(2)

df_QLCA["Person_GWP (341 Innenwände tragend (inkl. Aufzugschacht))"] = (
    df_QLCA["341 Innenwände tragend (inkl. Aufzugschacht)"] / denom
).round(2)

df_QLCA["Person_GWP (342 Innenwände)"] = (
    df_QLCA["342 Innenwände"] / denom
).round(2)

df_QLCA["Person_GWP (344 Innentüren)"] = (
    df_QLCA["344 Innentüren"] / denom
).round(2)

df_QLCA["Person_GWP (351 Decken)"] = (
    df_QLCA["351 Decken"] / denom
).round(2)

df_QLCA["Person_GWP (351 Treppe)"] = (
    df_QLCA["351 Treppe"] / denom
).round(2)

df_QLCA["Person_GWP (361 Dach)"] = (
    df_QLCA["361 Dach"] / denom
).round(2)

# --- KG 400 ---
df_QLCA["Person_GWP (KG 400)"] = (
    df_QLCA["GWP für Haustechnik (KG 400)"] / denom
).round(2)

df_QLCA["Person_GWP (410 Abwasser, Trinkwasser)"] = (
    df_QLCA["410 Abwasser, Trinkwasser"] / denom
).round(2)

df_QLCA["Person_GWP (420 Wärmeversorgung)"] = (
    df_QLCA["420 Wärmeversorgung"] / denom
).round(2)

df_QLCA["Person_GWP (430 Lüftung)"] = (
    df_QLCA["430 Lüftung"] / denom
).round(2)

df_QLCA["Person_GWP (440 Strom)"] = (
    df_QLCA["440 Strom"] / denom
).round(2)

df_QLCA["Person_GWP (460 Aufzug)"] = (
    df_QLCA["460 Aufzug"] / denom
).round(2)

# --- KG 500 ---
df_QLCA["Person_GWP (KG 500)"] = (
    df_QLCA["GWP für gesamte Geb.- Erschliessung (KG 500)"] / denom
).round(2)

df_QLCA["Person_GWP (500 Quartiersstrom)"] = (
    df_QLCA["500 Quartiersstrom"] / denom
).round(2)

df_QLCA["Person_GWP (500 Quartierswasserleitung)"] = (
    df_QLCA["500 Quartierswasserleitung"] / denom
).round(2)

df_QLCA["Person_GWP (510 Erdaushub Leitungen)"] = (
    df_QLCA["510 Erdaushub Leitungen"] / denom
).round(2)

df_QLCA["Person_GWP (510 Erdaushub Garagen)"] = (
    df_QLCA["510 Erdaushub Garagen"] / denom
).round(2)

df_QLCA["Person_GWP (521 Gehweg Radweg)"] = (
    df_QLCA["521 Gehweg Radweg"] / denom
).round(2)

df_QLCA["Person_GWP (522 Straße)"] = (
    df_QLCA["522 Straße"] / denom
).round(2)

df_QLCA["Person_GWP (524 Stellplätze)"] = (
    df_QLCA["524 Stellplätze"] / denom
).round(2)

df_QLCA["Person_GWP (524 Hochgaragen)"] = (
    df_QLCA["524 Hochgaragen"] / denom
).round(2)

df_QLCA["Person_GWP (524 Tiefgaragen)"] = (
    df_QLCA["524 Tiefgaragen"] / denom
).round(2)

df_QLCA["Person_GWP (Summe Betrieb & Nutzstrom)"] = (
    df_QLCA["Summe Betrieb & Nutzstrom"] / denom
).round(2)

# --- B6 ---
df_QLCA["Person_GWP (B6)"] = (
    df_QLCA["GWP für Wärmeversorgung (B6)"] / denom
).round(2)

df_QLCA["Person_GWP (GWP für Nutzstrom)"] = (
    df_QLCA["GWP für Nutzstrom"] / denom
).round(2)

df_QLCA.head()
```
