```python
"""
------------------------------------------------------------------------------
Exploratory tool for analysing and filtering the QLCA neighbourhood-level
dataset (6 ha) to identify and select scenarios for further investigation.

The VS Code interface allows the dataset to be filtered by six
scenario parameters and provides aggregated mean values for selected
indicators such as GWP, density and area.

Filter parameters:
- Typ / Building type:
    EFH – Einfamilienhaus / Single-family house
    DH  – Doppelhaus / Duplex
    RH  – Reihenhaus / Row house
    ZH  – Zeilenhaus / Linear building
    PH  – Punkthaus / Point building
    BW  – Block-Winkel / Block

- Dichte / Density:
    Hoch / High
    Mittel / Medium
    Gering / Low

- Erschließung (Sied.) / Road layout:
    Gitter / Grid
    Schlaufe / Curve
    Sackgasse / Cul-de-sac

- Bauart / Construction:
    Holz / Wood
    Massiv / Mineral

- Versorgungsart / Heat supply:
    Nahwärme kalt / Local energy
    Wärmepumpe / Heat pump
    Erdgas / Natural gas

- Energiebedarf / Energy demand:
    15 kWh
    30 kWh
    70 kWh

The filtered subset can be exported as a CSV file for further analysis.

The interface was used as an exploratory step in the workflow and is not
part of the final analysis results.
------------------------------------------------------------------------------
"""
```

```python
"""
------------------------------------------------------------------------------
Libraries
------------------------------------------------------------------------------
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown
from datetime import datetime
import os
```

```python
"""
------------------------------------------------------------------------------
Import data
------------------------------------------------------------------------------
"""

df_QLCA = pd.read_csv("../data/csv/QLCA_data.csv")

# Show all columns
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

df_QLCA.head()
```

```python
df_QLCA.dtypes
```

```python
df_QLCA["Obergeschosse"]= df_QLCA["Obergeschosse"].astype(float)
df_QLCA["A/V-Verhältnis Wohngebäude"]= df_QLCA["A/V-Verhältnis Wohngebäude"].round(1)
```

```python
"""
------------------------------------------------------------------------------
Dropdowns
------------------------------------------------------------------------------
"""


typ = widgets.Dropdown(
    options=["Alle"] + list(df_QLCA["Typ"].dropna().unique()),
    value="Alle",
    description="Typ"
)

bauart = widgets.Dropdown(
    options=["Alle"] + list(df_QLCA["Bauart"].dropna().unique()),
    value="Alle",
    description="Bauart"
)

energie = widgets.Dropdown(
    options = ["Alle"] + sorted(df_QLCA["Energiebedarf"].dropna().astype(str).unique()),
    value="Alle",
    description="Energiebedarf"
)


versorgung = widgets.Dropdown(
    options=["Alle"] + list(df_QLCA["Versorgungsart"].dropna().unique()),
    value="Alle",
    description="Versorgungsart"
)

dichte = widgets.Dropdown(
    options=["Alle"] + list(df_QLCA["Dichte"].dropna().unique()),
    value="Alle",
    description="Dichte"
)

erschliessung = widgets.Dropdown(
    options=["Alle"] + list(df_QLCA["Erschließung (Sied.)"].dropna().unique()),
    value="Alle",
    description="Erschließung (Sied.)"
)
```

```python
"""
------------------------------------------------------------------------------
Figures
------------------------------------------------------------------------------
"""

cols = {
    "Emissionen": [
        "GWP",
        "GWP für Bauwerk (KG 300)",
        "GWP für Haustechnik (KG 400)",
        "GWP für gesamte Geb.- Erschliessung (KG 500)",
        "GWP für Wärmeversorgung (B6)",
        "Person_GWP (GWP)",
        "GWP / m² NRF"
    ],
    "Dichte & Fläche": [
        "Quartiers-GFZ",
        "Quartiers-GRZ",
        "Nettoraumfläche NRF (m2)",
        "Verkehrsfläche gesamt m²",
        "BGF (m²) inkl. oberirdische Garagen",
        "BGF (m²) Wohnen (inkl. Keller)"
    ],
    
    "Sonstiges": [
        "Obergeschosse",
        "A/V-Verhältnis Wohngebäude",
        "Bewohner Kachel insg. (rechnerisch)"
    ]
}
```

```python
"""
------------------------------------------------------------------------------
Number formats
------------------------------------------------------------------------------
"""

format_dict = {
    "GWP": "{:.0f}",
    "GWP für Bauwerk (KG 300)": "{:.0f}",
    "GWP für Haustechnik (KG 400)": "{:.0f}",
    "GWP für gesamte Geb.- Erschliessung (KG 500)": "{:.0f}",
    "GWP für Wärmeversorgung (B6)": "{:.0f}",
    "Person_GWP (GWP)": "{:.0f}",
    "GWP / m² NRF": "{:.1f}",
    "Quartiers-GFZ": "{:.2f}",
    "Quartiers-GRZ": "{:.2f}",
    "Verkehrsfläche gesamt m²": "{:.2f}",
    "BGF (m²) inkl. oberirdische Garagen": "{:.2f}",
    "BGF (m²) Wohnen (inkl. Keller)": "{:.2f}",
    "Nettoraumfläche NRF (m2)": "{:.2f}",
    "Obergeschosse": "{:.1f}",
    "A/V-Verhältnis Wohngebäude": "{:.1f}",
    "Bewohner Kachel insg. (rechnerisch)": "{:.0f}"
}
```

```python
"""
------------------------------------------------------------------------------
Output
------------------------------------------------------------------------------
"""

out = widgets.Output()
current_result = None
```

```python
"""
------------------------------------------------------------------------------
Export-Button
------------------------------------------------------------------------------
"""


os.makedirs("../data/exports", exist_ok=True)

def clean(v):
    if v == "Alle":
        return None
    return str(v).replace(" ", "_")


def export_csv(b):
    if current_result is not None:

        ts = datetime.now().strftime("%Y%m%d_%H%M%S")

        parts = [
            clean(typ.value),
            clean(bauart.value),
            clean(energie.value),
            clean(versorgung.value),
            clean(dichte.value),
            clean(erschliessung.value)
        ]

        # filter All
        parts = [p for p in parts if p is not None]

        # file name
        filter_part = "_".join(parts) if parts else "GESAMT"

        filename = f"../data/exports/subset_{filter_part}_{ts}.csv"

        current_result.to_csv(filename, index=False)
        print(f"CSV exported: {filename}")

export_btn = widgets.Button(description="Export CSV")

export_btn.on_click(export_csv)
```

```python
"""
------------------------------------------------------------------------------
Filter function
------------------------------------------------------------------------------
"""

def filter_df(df):
    mask = pd.Series(True, index=df.index)

    if typ.value != "Alle":
        mask &= df["Typ"] == typ.value

    if bauart.value != "Alle":
        mask &= df["Bauart"] == bauart.value

    if energie.value != "Alle":
        mask &= df["Energiebedarf"] == energie.value

    if versorgung.value != "Alle":
        mask &= df["Versorgungsart"] == versorgung.value

    if dichte.value != "Alle":
        mask &= df["Dichte"] == dichte.value

    if erschliessung.value != "Alle":
        mask &= df["Erschließung (Sied.)"] == erschliessung.value

    return df[mask].copy()
```

```python
"""
------------------------------------------------------------------------------
Update function
------------------------------------------------------------------------------
"""

def update(*args):
    global current_result

    with out:
        out.clear_output()

        current_result = filter_df(df_QLCA)

        display(Markdown(f"### Ergebnis: n = {len(current_result)}"))

        for section, columns in cols.items():
            display(Markdown(f"## {section}"))

            existing_cols = [c for c in columns if c in current_result.columns]

            if len(existing_cols) > 0:
                display(
                    current_result[existing_cols]
                    .mean(numeric_only=True)
                    .to_frame("Mean")
                    .T
                    .style.format(format_dict)
                )

        

                
```

```python
"""
------------------------------------------------------------------------------
Observer
------------------------------------------------------------------------------
"""

for w in [typ, bauart, energie, versorgung, dichte, erschliessung]:
    w.style.description_width = "150px"
    w.layout.width = "400px"
    w.observe(update, "value")
```

```python
"""
------------------------------------------------------------------------------
Layout
------------------------------------------------------------------------------
"""

filter_widgets = [
    typ,
    bauart,
    energie,
    versorgung,
    dichte,
    erschliessung
]

for w in filter_widgets:
    w.observe(update, names="value")

ui = widgets.VBox([
    *filter_widgets,
    export_btn,
    out
])

display(ui)

update()



```
